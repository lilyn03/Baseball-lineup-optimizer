```python
class Player:
    def _init_(self, name, pa, s, d, t, hr, bb):
        self.name = name
        self.pa = pa
        self.s = s
        self.d = d
        self.t = t
        self.hr = hr
        self.bb = bb
        
```

```python
def determineSingleProb(player):
    p1 = ((player.s + player.bb)/player.pa)*100
    return p1
```

```python
def determineDoubleProb(player):
    p2 = (player.d/player.pa)*100
    return p2
```

```python
def determineTripleProb(player):
    p3 = (player.t/player.pa)*100
    return p3
```

```python
def determineHomeProb(player):
    p4 = (player.s/player.pa)*100
    return p4
```
